```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "City": ["New York", "San Francisco", "Los Angeles", "Chicago"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Drop the 'City' column
df = df.drop("City", axis=1)

print(df)
```

```python
# Create a DataFrame with null values
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, None, 35, 28],
    "City": ["New York", "San Francisco", None, "Chicago"]
}

df_null = pd.DataFrame(data)

# Fill null values
df_null["Age"] = df_null["Age"].fillna(df_null["Age"].mean())
df_null["City"] = df_null["City"].fillna("Unknown")

print(df_null)
```

```python
# Group by 'Category' and describe the 'Value' column

# Create the DataFrame
df_group = pd.DataFrame({
    "Category": ["A", "B", "A", "B", "A", "C"],
    "Value": [10, 20, 15, 25, 30, 35]
})
result = df_group.groupby("Category")["Value"].describe()

print(result)
```

```python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Concatenate df1 and df2 vertically
df_vertical = pd.concat([df1, df2], ignore_index=True)

# Merge df_vertical and df3 horizontally
df_horizontal = pd.concat([df_vertical, df3], axis=1)

print("Vertically concatenated DataFrame:")
print(df_vertical)

print("\nHorizontally merged DataFrame:")
print(df_horizontal)
```

```python
# Create a tuple of fruits
fruits = ("Apple", "Banana", "Orange")

# Create a set of numbers
numbers = {1, 2, 3}

# Try to add an element to the tuple
try:
    fruits.append("Mango")
except AttributeError:
    print("Cannot add to tuple because tuples are immutable.")

# Add an element to the set
numbers.add(4)

print("Tuple:", fruits)
print("Set:", numbers)
```

```python
# Create a dictionary of student names and scores
students = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78
}

# Update a student's score
students["Bob"] = 95

# Add a new student
students["David"] = 88

print(students)
```

```python
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# 1. Find the average salary by department
average_salary = df_employees.groupby('Department')['Salary'].mean()
print("Average salary by department:")
print(average_salary)

# 2. Get the names of employees with salary > 60000
high_salary_employees = df_employees[df_employees['Salary'] > 60000]['Name']
print("\nEmployees with salary > 60000:")
print(high_salary_employees)

# 3. Add a new Bonus column (10% of salary)
df_employees['Bonus'] = df_employees['Salary'] * 0.10

print("\nUpdated DataFrame:")
print(df_employees)
```
